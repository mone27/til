You can use `numpy.select` when you have multiple conditions to apply to an array.
It works like a vectorized `if/elif/else`, which is very convenient.

Small example:

```python
import numpy as np

x = np.array([2, 7, 12, 20, 30])

conditions = [x < 5, x < 15, x <= 20]
choices = ["small", "medium", "large"]

labels = np.select(conditions, choices, default="a real giant")
labels
```
